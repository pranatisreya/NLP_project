```python
import pandas as pd
from google.colab import files

def load_questions(csv_file):
  """Loads questions from a CSV file.

  Args:
      csv_file: Path to the CSV file.

  Returns:
      A pandas DataFrame containing questions as a column named 'prompt'
  """

  return pd.read_csv(csv_file)[['prompt']]

def split_and_save_csv(df, file_prefix, batch_size):
  """Splits a DataFrame into smaller CSV files based on batch size.

  Args:
      df: The DataFrame to split.
      file_prefix: The prefix for the output file names.
      batch_size: The number of rows per output CSV file.
  """

  num_batches = len(df) // batch_size + (1 if len(df) % batch_size else 0)

  for i in range(num_batches):
      start_index = i * batch_size
      end_index = min((i + 1) * batch_size, len(df))
      batch = df[start_index:end_index]

      output_file = f"{file_prefix}_{i+1}.csv"
      batch.to_csv(output_file, index=False)

  return num_batches

```

```python
import pandas as pd
import os
import json
import time
import requests
import openai
import copy
from functools import partial
from loguru import logger
from rich import print
from rich.console import Console
from rich.markdown import Markdown
from rich.prompt import Prompt
from datasets import Dataset
from datasets.utils.logging import disable_progress_bar

from tqdm import tqdm


DEBUG = int(os.environ.get("DEBUG", "0"))

def generate_together(
    model,
    messages,
    max_tokens=2048,
    temperature=0.7,
    streaming=False,
):
    output = None
    for sleep_time in [1, 2, 4, 8, 16, 32]:
        try:
            endpoint = "https://api.together.xyz/v1/chat/completions"
            if DEBUG:
                logger.debug(
                    f"Sending messages ({len(messages)}) (last message: `{messages[-1]['content'][:20]}...`) to `{model}`."
                )
            res = requests.post(
                endpoint,
                json={
                    "model": model,
                    "max_tokens": max_tokens,
                    "temperature": (temperature if temperature > 1e-4 else 0),
                    "messages": messages,
                },
                headers={
                    "Authorization": f"Bearer {os.environ.get('TOGETHER_API_KEY')}",
                },
            )
            if "error" in res.json():
                logger.error(res.json())
                if res.json()["error"]["type"] == "invalid_request_error":
                    logger.info("Input + output is longer than max_position_id.")
                    return None
            output = res.json()["choices"][0]["message"]["content"]
            break
        except Exception as e:
            logger.error(e)
            if DEBUG:
                logger.debug(f"Msgs: `{messages}`")
            logger.info(f"Retry in {sleep_time}s..")
            time.sleep(sleep_time)
    if output is None:
        return output
    output = output.strip()
    if DEBUG:
        logger.debug(f"Output: `{output[:20]}...`.")
    return output

def generate_together_stream(
    model,
    messages,
    max_tokens=2048,
    temperature=0.7,
):
    endpoint = "https://api.together.xyz/v1"
    client = openai.OpenAI(
        api_key=os.environ.get("TOGETHER_API_KEY"), base_url=endpoint
    )
    endpoint = "https://api.together.xyz/v1/chat/completions"
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature if temperature > 1e-4 else 0,
        max_tokens=max_tokens,
        stream=True,
    )
    return response

def generate_openai(
    model,
    messages,
    max_tokens=2048,
    temperature=0.7,
):
    client = openai.OpenAI(
        api_key=os.environ.get("OPENAI_API_KEY"),
    )
    for sleep_time in [1, 2, 4, 8, 16, 32]:
        try:
            if DEBUG:
                logger.debug(
                    f"Sending messages ({len(messages)}) (last message: `{messages[-1]['content'][:20]}`) to `{model}`."
                )
            completion = client.chat.completions.create(
                model=model,
                messages=messages,
                temperature=temperature,
                max_tokens=max_tokens,
            )
            output = completion.choices[0].message.content
            break
        except Exception as e:
            logger.error(e)
            logger.info(f"Retry in {sleep_time}s..")
            time.sleep(sleep_time)
    output = output.strip()
    return output

def inject_references_to_messages(
    messages,
    references,
):
    messages = copy.deepcopy(messages)
    system = f"""You have been provided with a set of responses from various open-source models to the latest user query.
    Your task is to synthesize these responses into a single, high-quality response. It is crucial to critically evaluate
    the information provided in these responses, recognizing that some of it may be biased or incorrect. Your response should
    not simply replicate the given answers but should offer a refined, accurate, and comprehensive reply to the instruction.
    Ensure your response is well-structured, coherent, and adheres to the highest standards of accuracy and reliability.

Responses from models:"""
    for i, reference in enumerate(references):
        system += f"\n{i+1}. {reference}"
    if messages[0]["role"] == "system":
        messages[0]["content"] += "\n\n" + system
    else:
        messages = [{"role": "system", "content": system}] + messages
    return messages

def generate_with_references(
    model,
    messages,
    references=[],
    max_tokens=2048,
    temperature=0.7,
    generate_fn=generate_together,
):
    if len(references) > 0:
        messages = inject_references_to_messages(messages, references)
    return generate_fn(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens,
    )

def process_fn(
    item,
    temperature=0.7,
    max_tokens=2048,
):
    references = item.get("references", [])
    model = item["model"]
    messages = item["instruction"]

    start_time = time.time()  # Record the start time

    output = generate_with_references(
        model=model,
        messages=messages,
        references=references,
        temperature=temperature,
        max_tokens=max_tokens,
    )

    end_time = time.time()  # Record the end time
    duration = end_time - start_time

    if DEBUG:
        logger.info(
            f"model: {model}, instruction: {item['instruction']}, output: {output[:20]}"
        )

    print(f"\nFinished querying [bold]{model}.[/bold]")

    return {"output": output, "time_taken": duration}

def process_question_moa(question, model_names,temperature, max_tokens):
    """Processes a single question using the MoA approach with multiple models.
    Args:question: The user's question to process.
        model_names: A list of model names to use.
    Returns:
        The final response generated by the MoA model.
    """
    # Prepare initial messages
    messages = [{"role": "user", "content": question}]
    # Initialize references
    references = [""] * len(model_names)
    # Generate responses from each model
    for i, model in tqdm(enumerate(model_names), desc="Generating Responses"):
        output = generate_with_references(
            model=model,
            messages=messages,
            references=references,
            temperature=temperature,
            max_tokens=max_tokens,
        )
        references[i] = output
    # Aggregate responses using the main model (you can choose one)
    main_model = model_names[0]  # Replace with your desired main model
    final_response = generate_with_references(
        model=main_model,
        messages=messages,
        references=references,
        temperature=0.7,
        max_tokens=2048,
    )
    return final_response


def main(csv_file, model_names):
    """Processes questions from a CSV and writes MoA results to the same file."""

    df = csv_file

    # Process each question and append the output to a list
    moa_results = []
    total_rows = len(df)
    start_time = time.time()

    with tqdm(total=total_rows, desc="Processing data") as pbar:
        for index, row in df.iterrows():
            question = row['prompt']
            output = process_question_moa(question, model_names, 0.7, 2048)
            print(f"Question: {question}")
            print(f"MoA Output: {output}")
            moa_results.append(output)
            pbar.update(1)
            # Estimate remaining time
            elapsed_time = time.time() - start_time
            remaining_time = (elapsed_time / (index + 1)) * (total_rows - index - 1)
            pbar.set_postfix(eta=f"{remaining_time:.1f}s")

    # Add MoA results as a new column
    df['MoA Results'] = moa_results

    # Overwrite the original CSV with the updated DataFrame
    df.to_csv(csv_file, index=False)

    print(f"MoA results written to '{csv_file}'.")




if __name__ == "__main__":

  model_names = [
      "Qwen/Qwen2-72B-Instruct",
      "Qwen/Qwen1.5-72B-Chat",
      "mistralai/Mixtral-8x22B-Instruct-v0.1",
      "databricks/dbrx-instruct",
  ]



```

```python

import pandas as pd
from tqdm import tqdm
def process_limited_data_moa(csv_file, model_names, batch_size=100):
    """Processes a limited number of rows from each batch of questions using MoA.
    Args:
        csv_file: Path to the CSV file.
        model_names: A list of model names to use.
        batch_size: The maximum number of rows to process per batch (default: 50).
    """
    df = pd.read_csv(csv_file)
    # Process data in batches
    with tqdm(total=len(df), desc=f"Processing {csv_file}") as outer_pbar:
        for i in range(0, len(df), batch_size):
            batch_df = df.iloc[i:i+batch_size]
            moa_results = []

            with tqdm(total=len(batch_df), desc="Processing batch", leave=False) as inner_pbar:
                for index, row in batch_df.iterrows():
                    question = row['prompt']
                    output = process_question_moa(question, model_names, 0.7, 2048)
                    print(f"Question: {question}")
                    print(f"MoA Output: {output}")
                    moa_results.append(output)
                    inner_pbar.update(1)

            # Combine MoA results with the original data (assuming 'prompt' column)
            batch_df['MoA Results'] = moa_results
            # Overwrite the original CSV with the updated batch (optional)
            batch_df.to_csv(csv_file, index=False)
            outer_pbar.update(len(batch_df))
            print(f"Finished processing batch of {len(batch_df)} rows from {csv_file}")
```

```python
import pandas as pd
import glob

def concatenate_csv_files(file_prefix, start_num, end_num):
  """Concatenates multiple CSV files into a single CSV file.

  Args:
      file_prefix: The prefix of the CSV files.
      start_num: The starting number of the CSV files.
      end_num: The ending number of the CSV files.
  """

  csv_files = []
  for i in range(start_num, end_num + 1):
      file_name = f"{file_prefix}_{i}.csv"
      csv_files.append(file_name)

  dfs = []
  for file in csv_files:
      df = pd.read_csv(file)
      dfs.append(df)

  combined_df = pd.concat(dfs, ignore_index=True)
  combined_df.to_csv("with_chatgpt_response_test_r1_claim_true_false_inconclusive_moa_result.csv", index=False)

# Example usage:
file_prefix = "with_chatgpt_response_test_r1_claim_true_false_inconclusive"
start_num = 1
end_num = 10
concatenate_csv_files(file_prefix, start_num, end_num)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, matthews_corrcoef, roc_auc_score, average_precision_score)

# Load the dataset
file_path = '/content/with_chatgpt_and_MOA_response_test_r1_claim_true_false_inconclusive.csv'
df = pd.read_csv(file_path)

# Define the columns
label_column = 'label'  # replace with the actual label column name
chatgpt_column = 'chatgpt_response'  # replace with the actual ChatGPT response column name
moa_column = 'MoA Results'  # replace with the actual MoA result column name

# Calculate confusion matrix for ChatGPT responses
chatgpt_cm = confusion_matrix(df[label_column], df[chatgpt_column])

# Flatten confusion matrix to get the sum of TP, FP, TN, FN across all classes
chatgpt_tp = np.diag(chatgpt_cm)
chatgpt_fp = chatgpt_cm.sum(axis=0) - chatgpt_tp
chatgpt_fn = chatgpt_cm.sum(axis=1) - chatgpt_tp
chatgpt_tn = chatgpt_cm.sum() - (chatgpt_tp + chatgpt_fp + chatgpt_fn)

# Calculate confusion matrix for MoA results
moa_cm = confusion_matrix(df[label_column], df[moa_column])

# Flatten confusion matrix to get the sum of TP, FP, TN, FN across all classes
moa_tp = np.diag(moa_cm)
moa_fp = moa_cm.sum(axis=0) - moa_tp
moa_fn = moa_cm.sum(axis=1) - moa_tp
moa_tn = moa_cm.sum() - (moa_tp + moa_fp + moa_fn)

# Print the confusion matrix values
print("ChatGPT Confusion Matrix Values (Sum Across All Classes):")
print(f"True Positives (TP): {chatgpt_tp.sum()}")
print(f"False Positives (FP): {chatgpt_fp.sum()}")
print(f"True Negatives (TN): {chatgpt_tn.sum()}")
print(f"False Negatives (FN): {chatgpt_fn.sum()}")

print("\nMoA Confusion Matrix Values (Sum Across All Classes):")
print(f"True Positives (TP): {moa_tp.sum()}")
print(f"False Positives (FP): {moa_fp.sum()}")
print(f"True Negatives (TN): {moa_tn.sum()}")
print(f"False Negatives (FN): {moa_fn.sum()}")

# Create a DataFrame for visualization of confusion matrix values
confusion_values_df = pd.DataFrame({
    'Model': ['ChatGPT', 'MoA'],
    'True Positives': [chatgpt_tp.sum(), moa_tp.sum()],
    'False Positives': [chatgpt_fp.sum(), moa_fp.sum()],
    'True Negatives': [chatgpt_tn.sum(), moa_tn.sum()],
    'False Negatives': [chatgpt_fn.sum(), moa_fn.sum()]
})

# Plot the confusion matrix values
plt.figure(figsize=(12, 8))
sns.barplot(x='variable', y='value', hue='Model',
            data=pd.melt(confusion_values_df, id_vars=['Model'], value_vars=['True Positives', 'False Positives', 'True Negatives', 'False Negatives']))
plt.title('Confusion Matrix Values Comparison')
plt.ylabel('Count')
plt.xlabel('Confusion Matrix Values')
plt.legend(title='Model')
plt.show()
```
